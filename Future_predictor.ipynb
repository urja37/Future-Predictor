```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from google.colab import files

# 1. Upload student_clean_full.csv to Google Colab
print("Please upload student_clean_full.csv:")
uploaded = files.upload()

# 2. Load dataset
df = pd.read_csv('student_clean_model.csv')

# 3. Select Target (G3) and Features
feature_cols = ['studytime', 'failures', 'absences', 'G1']
X = df[feature_cols]
y = df['G3']  # Final Grade

# 4. Train Linear Regression Model
model = LinearRegression()
model.fit(X, y)

print("\n Student Score Prediction Model trained successfully!")
```

```python
# 5. Live Prediction Function for User Input
def predict_student_score(studytime, failures, absences, g1):
    input_data = pd.DataFrame([[studytime, failures, absences, g1]], columns=feature_cols)
    prediction = model.predict(input_data)[0]
    # Cap prediction between 0 and 20 (standard grading scale)
    return max(0, min(20, prediction))

# Example Test: Studytime=3 (5-10 hrs), Failures=0, Absences=2, First Grade (G1)=14
sample_score = predict_student_score(studytime=3, failures=0, absences=2, g1=14)
print(f" Predicted Final Grade (G3): {sample_score:.2f} / 20")

# 6. Extract Feature Coefficients for Explainability
coef_df = pd.DataFrame({
    'Factor': ['Study Time', 'Past Failures', 'Absences', 'First Period Grade (G1)'],
    'Impact Coefficient': model.coef_
}).sort_values(by='Impact Coefficient', ascending=False)

print("\n Factors Influencing Predicted Score:")
print(coef_df.to_string(index=False))
```
